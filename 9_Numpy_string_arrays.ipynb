**Exercise 1: Creating string arrays**

```python
# 编写代码创建一个包含几个字符串的NumPy数组。
import numpy as np
array_string = np.array(['banan', 'jabko'])
print(array_string)
```

**Exercise 2: Concatenating string arrays**


```python
# 给定两个字符串数组 array1 和 array2，编写代码将它们拼接在一起。
import numpy as np
array_1 = np.array(['a', 'b', 'c'])
array_2 = np.array(['banana', 'apple'])
array_total = np.concatenate((array_1, array_2))
print(array_total)
```

**Exercise 3: Slicing string arrays**


```python
# 给定一个字符串数组 array，编写代码获取数组中索引从1到2的子集。
array_string = np.array(['a', 'b', 'c', 'd', 'e'])
array_sub = array_string[1:3]
print(array_sub)
```

**Exercise 4: Replacing elements in string arrays**


```python
# 给定一个字符串数组 array，编写代码将数组中所有的 "apple" 替换为 "orange"
import numpy as np
array_string = np.array(['banan', 'apple', 'cucumber', 'tomato'])
array_replace = np.where(array_string == 'apple', 'orange', array_string)
print(array_replace)
```

**Exercise 5: Case transformation in string arrays**

```python
# 给定一个字符串数组 array，编写代码将数组中的所有字符串转换为小写。
string_array = np.array(["APPLE","BANANA","CHERRY"])
string_lower = np.char.lower(string_array)
print(string_lower)
```

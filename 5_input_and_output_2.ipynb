**guessing the number game**

Write a program of guessing number, generate a integer range in 1 to 100 randomly. Ask the users to guess the number, and give the notifications

```python
import random

generated_number = random.randint(1, 100)

while True: # while true means that the program will run forever except it meets some situation to quit
  guess_number = int(input("guess a number between 1 to 100:"))

  if guess_number != generated_number:
    print("wrong")
  else:
    print("Right")
    break

# print always use with break, return only uses in the function


```

**Exercise 1: Calculating mean of NumPy arrays**


```python
# 给定一个包含若干整数的NumPy数组 array，编写代码计算这个数组的平均值。
import numpy as np
array_0 = np.array([1,2,3,4,5])
average = np.mean(array_0)
print(f"the average value of this array is :{average}")
```

**Exercise 2: Standard deviation of NumPy arrays**


```python
# 给定一个NumPy数组 array，编写代码计算这个数组的标准差。
import numpy as np
array_0 = np.random.rand(20)
standard_deviation = np.std(array_0)
print(f"the standard deviation of the array is {standard_deviation}")
```

**Exercise 3: Median of NumPy arrays**


```python
import numpy as np
array_0 = np.arange(5)
median_array_0 = np.median(array_0)
print(median_array_0)
```

**Exercise 4: Percentile of NumPy arrays**


```python
# 给定一个NumPy数组 array，编写代码计算这个数组的第25个和第75个分位数。
import numpy as np
array_0 = np.ones(5)
percentile_1 = np.percentile(array_0, 25)
percentile_3 = np.percentile(array_0, 75)
print(f"the 25 percentile is :{percentile_1}")
print(f"the 25 percentile is :{percentile_3}")
```

**Exercise 5: Minimum and maximum values of NumPy arrays**

```python
# 给定一个NumPy数组 array，编写代码找出这个数组的最小值和最大值。
import numpy as np
array_0 = np.array([1,2,3,4,5])
maximum = np.max(array_0)
minimum = np.min(array_0)
print(maximum)
print(minimum)
```

**Exercise 1: NumPy array reshaping**


```python
# 给定一个包含12个元素的一维NumPy数组 array，编写代码将其重塑为形状为 (3, 4) 的二维数组。
import numpy as np
array_0 = np.arange(12)
array_1 = array_0.reshape((3,4))
print(array_1)
```

**Exercise 2: Merging NumPy arrays**


```python
# 给定两个NumPy数组 array1 和 array2，编写代码将它们沿水平方向合并。
array_1 = np.array([1,2,3,4])
array_2 = np.ones(4)
array_concat = np.concatenate((array_1, array_2)) # combine array_1, array_2 as a tuple then put in
print(array_concat)
```

**Exercise 3: Splitting NumPy arrays**

```python
# 给定一个形状为 (6, 2) 的NumPy数组 array，编写代码将其分割为两个形状为 (3, 2) 的数组。
array_0 = np.arange(12).reshape((6,2))

split_arrays = np.split(array_0, 2)
print(split_arrays[0])
print(split_arrays[1])
```

**Exercise 4: Rotating NumPy arrays**


```python
# 给定一个形状为 (3, 3) 的NumPy数组 array，编写代码实现其矩阵的转置。
array_0 = np.arange(9).reshape(-1,3)
array_1 = array_0.T
print(array_1)
```

**Exercise 5: Summing NumPy arrays along axes**


```python
# 给定一个形状为 (4, 6) 的NumPy数组 array，编写代码沿着轴0（垂直方向）求和。
array_0 = np.random.rand(4,6)
total = np.sum(array_0, axis = 0)
print(total)
```

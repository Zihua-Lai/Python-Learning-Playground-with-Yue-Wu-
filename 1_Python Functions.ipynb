## Topic 2: String Repetition 

编写一个函数， 它接受一个字符串和一个整数作为参数， 并且返回字符串重复指定次数的结果

```python
string = "I love you"
rept = 12 

def repetition(string, rept):
    return string * rept 

print (repetition(string, rept))    # string * n  can repeat the string n times 
```

## Topic 3: String Counting

创建一个函数，它接受一个字符串为参数， 并且返回该字符串中每个字符出现的次数

```python
string = "love"

def counter(string):
    for letter in string:
        print(f"{letter} appears {string.count(letter)} times in this string")

print(counter(string))   # There is a count method for each string to count the letter appearance times 
```

```python
# solution 2 
def count_chars(s):
    char_count = {}
    for char in s:
        if char in char_count:
            char_count[char] +=1
        else:
            char_count[char] = 1
    return char_count

print(count_chars("love"))
```

## Topic 5: Finding Greatest Common Divisor (GCD)

创建一个函数，它接受两个整数作为参数，并且返回他们的最大公约数

```python
#the greatest common divisor (GCD) of two or more integers is the largest positive integer that divides each of the integers without leaving a remainder. 
# For example, the GCD of 8 and 12 is 4, since 4 is the greatest number that divides both 8 and 12 evenly. 

def gcb(a,b):
    while b:
        a, b = b, a % b
    return a 

print(gcb(48, 18))
```

## Topic 6: Fibonacci Sequence

编写一个函数， 它接受一个整数n 作为参数，并返回斐波那契数列的第n 项（从0开始计数）

```python
def fibonacci(n):
    if n == 0:
        return 0 
    elif n == 1:
        return 1 
    else:
        a,b = 0, 1
        for _ in range(2, n+1):
            a, b = b, a + b
        return b 
    
print(fibonacci(10))
```

## Topic 7:  Looking for the prime number 

创建一个函数，它接受一个整数作为参数， 并且返回这个数是否为素数

```python
def is_prime(num):
    if num <= 1 :
        return False 
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True 

print(is_prime(29))
```

## Topic 8: Sorting Algorithm Implementation

编写一个函数， 实现冒泡排序算法， 接受一个数字列表作为参数，并且返回排序后的列表

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j +1]:
                arr[j], arr[j +1] = arr[j +1], arr[j]
    return arr 
print(bubble_sort([64,34,15,37]))
```

**Exercise 1: Adding list elements**


```python
#编写一个函数，计算列表中所有数字的和。

def sum_list_elements(list):
  total = sum(list)
  return total

list_1 = [1,2,3,4,5]
print(sum_list_elements(list_1))
```

**Exercise 2: Deleting duplicated list elements**


```python
def remove_duplication_set(list_0):
    # 1. 将列表转换为集合 (set)：集合自动移除重复元素
    # 2. 将集合再转换回列表 (list)
    return list(set(list_0))

list_0 = [1,2,3,4,5,5]
print(remove_duplication_set(list_0))
```

**Exercise 3: Reverse the order of the  list elements**


```python
# 描述：编写一个函数，返回输入列表的逆序副本。
def reverse_list_order(list_0):
  list_reverse = list_0[::-1]
  return list_reverse
list_0 = [1,2,3,4,5]
print(reverse_list_order(list_0))
```

**Exercise 4: Sorting lists**

```python
# 编写一个函数，对输入的列表进行排序并返回排序后的列表。
def sorted_list(list_0):
  return sorted(list_0)

list_0 = [1,2,5,4,6,89,7]
print(sorted_list(list_0))
```

**Exercise 5: common area of 2 lists**

```python
# 编写一个函数，找出两个列表的共同元素，并返回这些元素组成的列表。
def common_area_list(list_1,list_2):
  list_0 = []
  for item in list_1:

      if item in list_2:
        list_0.append(item)
  return list_0
list_1 = [1,222,3,4,5]
list_2 = [2,3,4,5,6]

print(common_area_list(list_1, list_2))
```

**Topic 5: Array Element Reversal**

编写一个程序，使用循环语句反转一个列表中的元素顺序。


```python

def reverse_list(list_2):
  list_1 = []
  list_1 = list_2[::-1]
  return list_1



list_2 = [1,2,3,4]
print(reverse_list(list_2))


```

```python
def reverse_list(list_2):
  list_1 = list()
  for i in range(len(list_2)-1, -1, -1):
    list_1.append(list_2[i])
  return list_1

list_2 = [1,2,3,4]
print(reverse_list(list_2))

```

**Topic 6: Pattern Matching**

编写一个程序，使用循环语句检查一个字符串是否是另一个字符串的子串。

```python
def pattern_match(s1, s2):
  for i in range(len(s1)-len(s2)+1):
    if s1[i : i+len(s2)] == s2:
      return True
    else:
      return False

print(pattern_match("I love u", "you"))
```

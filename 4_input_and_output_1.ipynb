**Topic 2: Number Comparison**

编写一个程序，提示用户输入两个整数，并打印出较大数字


```python
num_1 = input("Please enter the first number:")
num_2 = input("Please enter the second number:")

if num_1 > num_2:
  print(num_1)
else:
  print(num_2)
```

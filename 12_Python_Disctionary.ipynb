```python
# 创建一个包含学生姓名和分数的字典，并打印出每个学生的分数。

dictionary_1 = {
    "Donna": 99,
    "Racheal": 89
}

for name, score in dictionary_1.items():
  print(f"{name} gets {score}")
```

**Exercise 3: Accumulate the values in a dictionary**


```python
# 给定一个包含多次考试分数的字典，编写一个函数计算学生的平均分。
import numpy as np
scores = {
    "first" : 20,
    "second": 30,
    "third": 50
    }
score_array= np.array(list(scores.values()))
print(np.mean(score_array))
```

**Exercise 4: Finding the maximum values by key**


```python
# 给定一个包含各种商品及其价格的字典，找出价格最高的商品。
product_price = {
    "apple": 20,
    "banana": 30,
    "milk": 15
}

highest_price = max(product_price.values())
product = max(product_price, key = product_price.get)

print(f"{product} has the highest_price, which is {highest_price}")
```

**Exercise 5: Nested Dictionary**

```python
# 给定一个嵌套字典，该字典包含学生信息，包括姓名、分数和年龄。编写一个函数打印每个学生的年龄和平均分

def calculate_average(scores_list):
  total = sum(scores_list)
  return total / len(scores_list)

# 打印学生信息的函数，包括计算平均分
def print_student_info(students):
  for student_id, info in students.items():
    name = info['name']
    scores = info['scores']
    age = info['age']
    average_score = calculate_average(scores)
    print(f"学生姓名: {name}, 平均分: {average_score:.2f}, 年龄: {age}")
# 给定的嵌套字典students_info
students_info = {
'001': {
'name': 'Alice',
'scores': [88, 92, 79],
'age': 16
},
'002': {
'name': 'Bob',
'scores': [75, 81, 90],
'age': 17
}
}
# 调用函数打印每个学生的年龄和平均分
print_student_info(students_info)
```

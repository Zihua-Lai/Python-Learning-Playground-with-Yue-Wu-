**Topic 1: Print Multiplication Table**

编写一个程序，使用循环语句打印1到10的乘法表

```python
print("The multiplication table")
for i in range(1,11):
  for j in range(1, i+1):
    print(f"{i}* {j} = {i* j}", end = "")
  print()

```

**Topic 2: Number Summation**

编写一个函数，使用循环语句计算1到n的数字之和。

```python
def summation(n):
  list_1 = list()
  for i in range(1, n+1):
    list_1.append(i)
  return sum(list_1)

print(summation(3))
```

**Topic 3: Febonacci Sequence**
* What is Febonacci Sequence?  --> 斐波那契数列是一个每一项都是前两项和的序列

编写一个程序，使用循环语句打印前n项斐波那契数列。

```python
def febonacci(n):
  a, b = 0, 1
  for i in range(n):
    print(a, end = "")
    a, b = b, a+b

febonacci(3)

```

**Topic 4: Factorial Calculation**

编写一个函数，使用循环语句计算一个非负整数的阶乘。

---



```python
def factorial(n):
  result = 1
  for i in range(1, n+1):
    result *= i
  return result

print(factorial(2))

```
